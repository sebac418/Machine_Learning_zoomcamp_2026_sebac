```python
# For the most frequent value we use:
print(df["horsepower"].mode()) # dataframe
frequent_number = df["horsepower"].mode()[0] # Float Number
```

```python
# Now we again calculate the median value:
df["horsepower"].fillna(frequent_number).median()
```

```python
# Now we extract only the cars form asia:

df_asia = df[df["origin"] == "Asia"]
df_asia.head()
```

```python
df_asia_cut = df_asia[["vehicle_weight", "model_year"]]
df_asia_cut.head()
```

```python
df_asia_cut_2 = df_asia_cut.head(7)
print(df_asia_cut_2)
```

```python
# We creat the 2-dimensional array:
X = df_asia_cut_2.to_numpy()
print(X)
```

```python
XTX = np.dot(X.T, X)
print(XTX)
```

```python
inverse = np.linalg.inv(XTX)
print(inverse)
```

```python
y= np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
temp = np.dot(inverse, X.T)
print(temp)
w=np.dot(temp, y)
print(w)

print(np.sum(w))
```
